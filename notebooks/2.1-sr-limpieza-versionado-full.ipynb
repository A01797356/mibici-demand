````python
%load_ext autoreload
%autoreload 2
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
````

````python
from mibici_demand.config import CLEANING_PARAMS, RAW_TRIPS_DIR
from mibici_demand.dataset import TripCleaner, clean_files, load_stations

PATTERN = "datos_abiertos_*.csv"  # periodo de trabajo
files = sorted(RAW_TRIPS_DIR.glob(PATTERN))
stations = load_stations()
cleaner = TripCleaner(station_ids=set(stations["station_id"]) if stations is not None else None)
CLEANING_PARAMS
````

````python
trips = clean_files(files, cleaner)
trips.info()
````

## Reporte de limpieza (filas eliminadas por regla)

````python
report = pd.Series(cleaner.report_)
report
````

````python
steps = report.drop(["rows_raw", "rows_clean", "invalid_age_set_nan"], errors="ignore")
(100 * steps / report["rows_raw"]).round(3).sort_values().plot.barh(
    title="% de viajes eliminados por regla", figsize=(7, 3));
````

## Validaciones del resultado

````python
assert trips["start_time"].notna().all()
assert (trips["duration_min"] >= CLEANING_PARAMS["min_duration_min"]).all()
assert (trips["duration_min"] <= CLEANING_PARAMS["max_duration_h"] * 60).all()
trips.describe()
````
